## 1. Static Method :

```python
# 1. Basic Static Method :
class Calculator:
  @staticmethod
  def add(a,b):
    return a+b

print(Calculator.add(5,6))
```

```python
# 2. Validation using Static Method :
class Student:
  @staticmethod
  def is_valid_marks(marks):
    if 0<=marks<=100:
      return True
    else:
      return False

print(Student.is_valid_marks(88))
print(Student.is_valid_marks(101))
```

```python
# 3. Real-world Employee Utility :
class Employee:
  @staticmethod
  def calculate_bonus(salary,percentage):
    bonus = salary * percentage / 100
    return f"Bonus : {bonus}"

print(Employee.calculate_bonus(50000 , 10))
```

## 2. Class Method :

```python
# 1. Basic Class Method :
class Student:
  school = "ABC School"

  @classmethod
  def show_school(cls):
    print(cls.school)

Student.show_school()
```

```python
# 2. Class Variable Update :
class Employee:
  company = "ABC Ltd"

  @classmethod
  def change_company(cls , new_comapny):
    cls.company = new_comapny
    print(cls.company)

Employee.change_company("XYZ ltd")
```

```python
# 3. Alternative Constructor :
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def from_string(cls, data):
        name, age = data.split(",")
        return cls(name, int(age))


s = Student.from_string("Krish,20")

print(s.name)
print(s.age)
```
