### 1.File handling :

```python
# 1. Write to a File
"""
Expected file content:
Name: Abc
Age: 21
"""
with open("student.txt","w") as f:
  name = input("Enter your name : ")
  age = int(input("Enter your age : "))

  text = f"Name : {name}\nAge: {age}"
  f.write(text)
```

```python
# 2. Read a File :
with open("student.txt","r") as f:
  text = f.read()
  print(text)
```

```python
# 3. Add Multiple Students :
with open("students.txt","w") as f:
  n = 5
  for i in range(n):
      name = input(f"Enter name of student : ")
      text = f"Name : {name}\n"
      f.write(text)
```

```python
with open("students.txt","r") as f:
  text = f.read()
  print(text)
```

```python
# 4. Count Lines :
with open("students.txt","r") as f:
  text = f.readlines()
  count = 0
  for line in text:
    count += 1
  print("Total students :",count)

```

```python
with open("students.txt", "r") as f:
  search_name = input("Enter name : ")
  found = False
  for line in f:
    line = line.strip()
    if search_name in line:
      found = True
      break

  if found:
    print("Student found")
  else:
    print("Student not found")
```

### 2. Exception Handling :

```python
# 1. Safe Division :
try :
  num1 = int(input("Enter a number : "))
  num2 = int(input("Enter a number : "))
  division = num1/num2
  print(f"Division :{division}")
except ZeroDivisionError:
  print("Cannot divide by 0")
except ValueError:
  print("Invalid input")
```

```python
# 2. Invalid Number Input :
try:
  num1 = int(input("Enter a number : "))
  num2 = int(input("Enter a number : "))
  total = num1+num2
  print(total)
except ValueError:
  print("Please enter a valid number")
```

```python
# 3. File Handling + Exception :
try:
  with open("Students.txt","r") as f:
    print("Found")
except FileNotFoundError:
  print("File not found")
```

```python
# 4. Student Marks Validation :
try:
  marks = int(input("Enter your marks : "))
  if marks<0 or marks>100:
    print("Marks must be between 0 and 100")
  else:
    print(marks)
except ValueError:
  print("Invalid marks")
```

```python
# 5. Safe Student Search :
try:
  with open("students.txt","r") as f:
    search_student = input("Enter name : ")

    found = False
    for line in f:
      line = line.strip()
      if search_student in line:
        found = True
        break
    if found:
      print("Student found")
    else:
      print("Student not found")
except FileNotFoundError:
  print("Student file not found")
```
