### 1. List :

```python
# 1. Sum of List :
numbers = [5,8,10,2]

print(sum(numbers))
```

```python
# 2. Find Maximum:
numbers = [15,12,18,20]
highest = numbers[0]

for num in numbers:
    if num > highest:
        highest = num
print(highest)
```

```python
# 3. Remove Duplicates :
numbers = [10, 20, 10, 30, 20, 40, 30]
unique_numbers = list(set(numbers))
print(unique_numbers)
```

```python
# 4. Count Even Numbers :
numbers = [12, 7, 8, 15, 20, 33, 42]
count=0
for num in numbers:
  if num%2==0:
    count+=1
print(count)
```

```python
# 5. Reverse List :
numbers = [1, 2, 3, 4, 5]
numbers.reverse()
print(numbers)
```

```python
# 6. Find Minimum :
numbers = [45, 12, 89, 23, 7, 56]
minimum = numbers[0]
for num in numbers:
  if num<minimum:
    minimum = num
print(minimum)
```

```python
# 7. Count Odd Numbers :
numbers = [12, 7, 8, 15, 20, 33, 42, 9]
count = 0
for num in numbers:
  if num%2 != 0:
    count+=1
print(count)
```

```python
# 8. Remove Negative Numbers :
numbers = [10, -5, 20, -8, 15, -2, 30]
new_list = []
for num in numbers:
  if num>0:
    new_list.append(num)
print(new_list)


```

```python
# 9. Find Second Largest :
numbers = [10, 25, 8, 40, 15, 30]
largest = numbers[0]
second_largest = numbers[0]
for num in numbers:
  if num>largest:
    second_largest = largest
    largest = num
  elif num>second_largest and num != largest:
    second_largest = num
print(second_largest)
```

```python
# 10. Calculate Average :
numbers = [10, 20, 30, 40, 50]
numbers_sum = sum(numbers)
avg = numbers_sum / len(numbers)
print(avg)
```

### 4. Dictionary :

```python
# 1. Student Details :
student = {
    "name": "Rahul",
    "age": 21,
    "marks": 85
}
print(student)
```

```python
# 2. Add New Key :
student = {
    "name": "Rahul",
    "marks": 85
}
student["grade"] = "A"
print(student)
```

```python
# 3. Update Value :
student = {
    "name": "Rahul",
    "marks": 75
}
student["marks"] = 90
print(student)
```

```python
# 4. Find Highest Salary :
employees = {
    "Rahul": 50000,
    "Aman": 75000,
    "Ravi": 60000
}

highest_salary = 0
highest_employee = ""

for name, salary in employees.items():
    if salary > highest_salary:
        highest_salary = salary
        highest_employee = name

print(highest_employee)
print(highest_salary)
```

```python
# 5. Dictionary Loop :
student = {
    "name": "Rahul",
    "age": 21,
    "course": "Python",
    "marks": 88
}

for key, value in student.items():
  print(f"{key} : {value}")
```

### 5. List Comprehension :

```python
# 1. Squares :
numbers = [1, 2, 3, 4, 5]
squares = [x**2 for x in numbers]
print(squares)
```

```python
# 2. Even Numbers :
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
even = [x for x in numbers if x%2==0]
print(even)
```

```python
# 3. Numbers Greater Than 50 :
numbers = [20, 65, 40, 80, 35, 90]
new_numbers = [x for x in numbers if x>50]
print(new_numbers)
```

```python
# 4. Convert Strings to Uppercase :
names = ["rahul", "aman", "ravi", "neha"]
upper_names = [x.upper() for x in names]
print(upper_names)
```

```python
# 5. Pass Students :
students = [
    {"name": "Rahul", "marks": 85},
    {"name": "Aman", "marks": 35},
    {"name": "Ravi", "marks": 72},
    {"name": "Neha", "marks": 28}
]
pass_students = [student["name"] for student in students if student["marks"]>=40]
print(pass_students)
```
