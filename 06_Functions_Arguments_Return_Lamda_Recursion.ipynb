```python
# 3. Create a function calculator() that performs:
print("Choose operation to perform : + , - , * , /")
def calculator(a,b,operator):
  if operator == "+":
    return (f"Addition is : {a+b}")
  elif operator == "-":
    return (f"Substraction is : {a-b}")
  elif operator == "*":
    return (f"Multiplication is : {a*b}")
  elif operator == "/":
    if b == 0:
      return (f"A number cannot divide by 0")
    else:
      return (f"Divid is : {a/b}")
c = calculator(5,0,"*")
print(c)
```

```python
# 4. Create a function student_result() that takes marks of 5 subjects and calculates:
def student_result(sub_1,sub_2,sub_3,sub_4,sub_5):
  Total_marks = 500
  Total = sub_1 + sub_2 + sub_3 + sub_4 + sub_5
  print(f"Your Total marks is : {Total}")
  Percentage = (Total * 100) / Total_marks
  print(f"Your Percentage is : {Percentage}%")
  if Percentage >= 90 :
    return ("Grade : A")
  elif Percentage<=80:
    return ("Grade : B")
  elif Percentage<=60:
    return ("Grade : C")
  else:
    return ("Grade : F(Fail)")
s = student_result(98,95,87,90,91)
print(s)
```

```python
# 5. Create a function calculate_bill() for a shopping bill.
"""
Take purchase amount.
Apply discount based on amount.
Calculate final amount.
Display the result.
"""
def calculate_bill(amount):
  if amount >= 2000 and amount < 4000:
    discount = amount * 0.10
  elif amount >=4000 and amount < 6000:
    discount = amount * 0.20
  elif amount >=6000 and amount < 8000:
    discount = amount * 0.30
  else:
    discount = amount * 0.40
  final_amount = amount - discount
  return amount , discount , final_amount
amount = int(input("Enter your amount : "))
amount , discount,final_amount = calculate_bill(amount)
print(f"Amount : {amount}")
print(f"Discount : {discount}")
print(f"final amount : {final_amount}")
```

### 5.Recursion :

```python
# 1. Countdown:
def countdown(n):
  if n == 0:
    return
  print(n)
  countdown(n-1)
countdown(5)
```

```python
# 2. Sum 1 to N :
def sum_n(n):
  if n == 0:
    return 0
  return n + sum_n(n-1)
s = sum_n(5)
print(s)


```

```python
# 3. Factorial:
def Factorial(n):
  if n == 0 or n==1 :
    return 1
  return n * Factorial(n-1)
f = Factorial(5)
print(f)
```

```python
# 4. Fibonacci:
def fibonacci(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(6))
```

```python
# 5. reverse a string :
def reverse_string(s):
    if s == "":
        return ""
    return reverse_string(s[1:]) + s[0]

print(reverse_string("hello"))
```

### 1 : Student Result System :

```python
def calculate_total(marks):
  return sum(marks)

def calculate_percentage(total,subjects):
  return total/subjects

def calculate_grade(percentage):
    if percentage >= 90:
        return "A"
    elif percentage >= 80:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 40:
        return "D"
    else:
        return "F"
def result_checker(marks):
  for marks in marks:
    if marks<40:
      return "Fail"
  return "Pass"

n = int(input("How many students? "))

students = []

for i in range(n):

    name = input("Enter student name: ")

    marks = []

    for j in range(5):
        mark = int(input(f"Enter marks of subject {j + 1}: "))
        marks.append(mark)

    total = calculate_total(marks)
    percentage = calculate_percentage(total, len(marks))
    grade = calculate_grade(percentage)
    result = result_checker(marks)

    student = {
        "name": name,
        "marks": marks,
        "total": total,
        "percentage": percentage,
        "grade": grade,
        "result": result
    }

    students.append(student)

students.sort(key=lambda x:x["percentage"],reverse=True)

print("\n=====ranking=====")
for student in students:
  print(
    student["name"],
    student["percentage"],
    student["grade"]
  )
```

### 2 : Employee Payroll System :

```python
def total_salary(basic,bonus,deduction):
  total_salary = basic + bonus - deduction
  return total_salary

def salary_grade(salary):
  if salary >= 80000:
    return "A"
  elif salary >= 50000:
    return "B"
  elif salary >= 30000:
    return "C"
  else:
    return "D"

def calculate_tax(salary):
  if salary >= 100000:
    return salary * 0.10
  elif salary >= 50000:
    return salary *.05
  else:
    return salary * 0

def calculate_net_salary(salary,tax):
  net_salary = salary- tax
  return net_salary


n = int(input("How many employees ? : "))
employees = []

for i in range(n):
  name = input("enter your name : ")
  basic_salary = float(input("Enter basic salary: "))
  bonus = float(input("Enter bonus: "))
  deduction = float(input("Enter deduction: "))

  salary = total_salary(basic_salary,bonus,deduction)
  grade = salary_grade(salary)
  tax = calculate_tax(salary)
  net_salary = calculate_net_salary(salary,tax)

  employee = {
    "name": name,
    "basic": basic_salary,
    "bonus": bonus,
    "deduction": deduction,
    "total_salary": salary,
    "grade": grade,
    "tax": tax,
    "net_salary": net_salary
  }
  employees.append(employee)

employees.sort(key=lambda x:x["total_salary"],reverse=True)


print("\n=====employee ranking by salary=====")
for employee in employees:
  print(
    employee["name"],
    "-->",
    employee["total_salary"],
    "-->",
    employee["grade"],
    "-->",
    employee["net_salary"]
  )
```
