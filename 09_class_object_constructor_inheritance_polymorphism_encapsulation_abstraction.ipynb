## 3. Inheritance :

```python
# 1. Single Inheritance: Employee → Manager :
class Employee:
  def __init__(self,name,salary):
    self.name = name
    self.salary = salary

  def display_employee(self):
    print(f"Name : {self.name}\nSalary : {self.salary}")

class Manager(Employee):
  def __init__(self,name,salary,department):
    super().__init__(name,salary)
    self.department = department

  def display_manager(self):
    print(f"Department : {self.department}")

m = Manager("Rohan",45000,"IT")

m.display_employee()
m.display_manager()
```

```python
# 2. Single Inheritance: Vehicle → Car :
class Vehicle:
  def __init__(self,brand,speed):
    self.brand = brand
    self.speed = speed

  def show_vehicle(self):
    print(f"Brand : {self.brand}\nSpeed : {self.speed}")

class Car(Vehicle):
  def __init__(self,brand,speed,fuel_type):
    super().__init__(brand,speed)
    self.fuel_type = fuel_type

  def show_car(self):
    print(f"Fuel Type : {self.fuel_type}")


c = Car("Toyota",80,"Petrol")
c.show_vehicle()
c.show_car()
```

```python
# 3. Multilevel Inheritance: Person → Student → Graduate :
class Person:
  def __init__(self,name):
    self.name = name

  def show_name(self):
    print(f"Name : {self.name}")

class Student(Person):
  def __init__(self,name,course):
    super().__init__(name)
    self.course = course

  def show_course(self):
    print(f"Course : {self.course}")

class Graduate(Student):
  def __init__(self,name,course,degree):
    super().__init__(name,course)
    self.degree = degree

  def show_degree(self):
    print(f"Degree : {self.degree}")

g = Graduate("Rohan","Information Technology","B.Tech")
g.show_name()
g.show_course()
g.show_degree()
```

```python
# 4. Multilevel Inheritance: Animal → Dog → Puppy :
class Animal:
  def eat(self):
    print(f"Animal is eating")

class Dog(Animal):
  def bark(self):
    print(f"Dog is barking")

class Puppy(Dog):
  def play(self):
    print("Puppy is playing")

p = Puppy()
p.eat()
p.bark()
p.play()
```

```python
# 5. Multiple Inheritance: Teacher + Researcher → Professor :
class Teacher:
  def teach(self):
    print("Teacher is teaching")

class Researcher:
  def research(self):
    print("Researcher is doing research")

class Professor(Teacher,Researcher):
  def lecture(self):
    print("Professor is giving lecture")

p = Professor()
p.teach()
p.research()
p.lecture()
```

## 4. Polymorphism :

```python
# 1. Same Method, Different Classes :
class Dog:
  def sound(self):
    print("Dog barks")

class Cat:
  def sound(self):
    print("Cat meows")

d = Dog()
c = Cat()

d.sound()
c.sound()
```

```python
# 2. Method Overriding :
class Animal:
  def sound(self):
    print("Sound of animal")

class Dog(Animal):
  def sound(self):
    print("Dog Barks")

class Cat(Animal):
  def sound(self):
    print("Cat meows")

d = Dog()
c = Cat()

d.sound()
c.sound()
```

```python
# 3. Common Function :
class Car:
  def start(self):
    print("Car started")

class Bike:
  def start(self):
    print("Bike started")

class Bus:
  def start(self):
    print("Bus started")

def start_vehicle(vehicle):
  vehicle.start()

start_vehicle(Car())
start_vehicle(Bike())
start_vehicle(Bus())

```

```python
# 5. Payment System :
class UPI:
  def pay(self,amount):
    print(f"Paid {amount} using UPI")

class Card:
  def pay(self,amount):
    print(f"Paid {amount} using card")

class Cash:
  def pay(self,amount):
    print(f"Paid {amount} using cash")

def payment(payment_methods):
  payment_methods.pay(500)

payment(UPI())
payment(Card())
payment(Cash())
```

## 5. Encapsulation :

### 1. Access Modifiers :

```python
# 1. Public Attribute :
class Students:
  def __init__(self,name,age,course):
    self.name = name
    self.age = age
    self.course = course

s = Students("Rohan",20,"B.Tech")
print(s.name)
print(s.age)
print(s.course)

s.name = "Rahul"
print(s.name)
```

```python
# 2. Protected Attribute :
class Employee:
  def __init__(self,name,salary):
    self.name = name
    self._salary = salary

  def show_salary(self):
    print(f"Salary : {self._salary}")

e = Employee("Rohan",45000)
print(e.name)

e.show_salary()
```

```python
# 3. Private Attribute :
class BankAccount:
  def __init__(self,balance):
    self.__balance = balance

  def show_balance(self):
    print(f"Balance : {self.__balance}")

b = BankAccount(45000)
b.show_balance()
```

### 2. Getter and setter :

```python
# 1. Student Name :
class Student:
  def __init__(self,name):
    self.__name = name

  def get_name(self):
    return f"Name : {self.__name}"

  def set_name(self,name):
    self.__name = name


s = Student("Rohan")
print(s.get_name())

s.set_name("Rahul")
print(s.get_name())
```

```python
# 2. Student Marks :
class Student:
  def __init__(self,marks):
    self.__marks = marks

  def get_marks(self):
    return f"Marks : {self.__marks}"

  def set_marks(self,marks):
    if 0<=marks<=100:
      self.__marks = marks
      print(f"Marks : {self.__marks}")
    else:
      print("Enter valid marks")

s = Student(80)
print(s.get_marks())

s.set_marks(101)
print(s.get_marks())
```

### 3. @property :

```python
# 1. Student Marks :
class Student:
  def __init__(self,marks):
    self.__marks = marks

  @property
  def marks(self):
    return f"Marks : {self.__marks}"

  @marks.setter
  def marks(self,marks):
    if 0<=marks<=100:
      self.__marks = marks
    else:
      print("Invalid marks")

s =Student(80)
print(s.marks)

s.marks = 101
print(s.marks)
```

```python
# 2. Bank Account :
class BankAccount:
  def __init__(self,balance):
    self.__balance = balance

  @property
  def balance_show(self):
    return f"Balance : {self.__balance}"

  @balance_show.setter
  def balance_show(self,balance):
    if balance >=0:
      self.__balance = balance
    else:
      print("Invalid balance")

b = BankAccount(5000)
print(b.balance_show)

b.balance_show = -1
print(b.balance_show)
```

```python
# 3. Employee Salary :
class Employee:
  def __init__(self,salary):
    self.__salary = salary

  @property
  def salary_show(self):
    return f"Salary : {self.__salary}"

  @salary_show.setter
  def salary_show(self,salary):
    if salary>=0:
      self.__salary = salary
    else:
      print("Invalid salary")

e = Employee(60000)
print(e.salary_show)

e.salary_show = -34000
print(e.salary_show)
```

## 6. Abstraction :

```python
# 1. Animal :
from abc import ABC , abstractmethod
class Animal(ABC):
  @abstractmethod
  def sound(self):
    pass

class Dog(Animal):
  def sound(self):
    print("Dog barks")

class Cat(Animal):
  def sound(self):
    print("Cat meows")

d = Dog()
c = Cat()

d.sound()
c.sound()
```

```python
# 2. Vehicle :
from abc import ABC , abstractmethod
class Vehicle(ABC):
  @abstractmethod
  def start(self):
    pass

class Car(Vehicle):
  def start(self):
    print("Car starts with key")

class Bike(Vehicle):
  def start(self):
    print("Bike starts with self-start")

C = Car()
B = Bike()

C.start()
B.start()
```

```python
# 3. Payment :
from abc import ABC , abstractmethod
class Payment(ABC):
  @abstractmethod
  def pay(self,amount):
    pass

class UPI(Payment):
  def pay(self,amount):
    print(f"{amount} Paid using UPI")

class Card(Payment):
  def pay(self,amount):
    print(f"{amount} Paid using Card")

class Cash(Payment):
  def pay(self,amount):
    print(f"{amount} Paid using Cash")


U = UPI()
C = Card()
C1 = Cash()

U.pay(3000)
C.pay(2000)
C1.pay(1000)
```

```python
# 4. Employee :
from abc import ABC , abstractmethod
class Employee(ABC):
  @abstractmethod
  def calculate_salary(self):
    pass

class FullTimeEmployee(Employee):
  def __init__(self,salary):
    self.salary = salary
  def calculate_salary(self):
    print(f"Salary : {self.salary}")

class PartTimeEmployee(Employee):
  def __init__(self,hour,hourly_rate):
    self.hour = hour
    self.hourly_rate = hourly_rate
  def calculate_salary(self):
    PartTime_salary = self.hour * self.hourly_rate
    print(f"Salary : {PartTime_salary}")

f = FullTimeEmployee(50000)
p = PartTimeEmployee(5,500)

f.calculate_salary()
p.calculate_salary()
```

```python
# 5. Notification System :
from abc import ABC , abstractmethod
class Notification(ABC):
  @abstractmethod
  def send(self,message):
    pass

class EmailNotification(Notification):
  def send(self,message):
    print(f"Message : {message} ")


class SMSNotification(Notification):
  def send(self,message):
    print(f"Message : {message}")


class WhatsAppNotification(Notification):
  def send(self,message):
    print(f"Message : {message}")

e = EmailNotification()
s = SMSNotification()
w = WhatsAppNotification()


e.send("For Email")
s.send("For SMS")
w.send("For Whatsapp")
```
